```python
# Reseaux de neurones avec 2 noeuds

import numpy as np 

# Paramètres de base 
input_data = np.array([3, 5])
weights = {"node_0_0" : np.array([2, 4]),
            "node_0_1" : [4, -5],
            "node_1_0" : [-1, 2],
            "node_1_1" : [1, 2],
            "output" : [2, 7]}

# Fonction ReLU
def relu(input):
    if input < 0:
        output = 0
    else: output = input 
    return output

# Calcule de la première couche
node_0_0_input = (input_data * weights["node_0_0"]).sum()
node_0_0_output = relu(node_0_0_input)
node_0_1_input = (input_data * weights["node_0_1"]).sum()
node_0_1_output = relu(node_0_1_input)

# Calcule de la deuxième couche
hidden_0_outputs = np.array([node_0_0_output, node_0_1_output])
node_1_0_input = (hidden_0_outputs * weights["node_1_0"]).sum()
node_1_0_output = relu(node_1_0_input)
node_1_1_input = (hidden_0_outputs * weights["node_1_1"]).sum()
node_1_1_output = relu(node_1_1_input)

# Calcule model_output
hidden_1_outputs = np.array([node_1_0_output, node_1_1_output])
modele_output = (hidden_1_outputs * weights["output"]).sum()



print(node_1_0_input)
print(node_1_0_output)
print(node_1_1_input)
print(node_1_1_output)
print(modele_output)
```

```python
import pandas as pd 

#Load the data 

sp500_data = pd.read_csv("sp500.csv", index_col='Date', parse_dates=True)
print(sp500_data.head())
```

```python
import matplotlib.pyplot as plt 

plt.plot(sp500_data['Close'], color='red', label='sp500', linewidth=1)
plt.title("Daily Close price", fontstyle='italic')
plt.xlabel('Date')
plt.ylabel('Prix $ ')
plt.grid(True, which='both', linestyle=':')
plt.legend()

plt.show()
```

```python
import plotly.graph_objects as go 

# Define the candlestick
candlestick = go.Candlestick(
    x = sp500_data.index,
    open = sp500_data['Open'],
    high = sp500_data['High'],
    low = sp500_data['Low'],
    close = sp500_data['Close']
)

# Create a plot 
fig = go.Figure(data=[candlestick])

# Show the plot
fig.show()
fig.write_image("candlestick.png")
```
