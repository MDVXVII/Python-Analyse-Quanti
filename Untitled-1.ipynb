```python
# Deep Learning course

import numpy as np

input_data = np.array([2, 3])
input_data_2 = np.array([])
weights = {"node_0" : np.array([1, 1]),
           "node_1" : np.array([-1, 1]),
           "output" : np.array([2, -1])}
node_0_value = (input_data * weights["node_0"]).sum()
node_1_value = (input_data * weights["node_1"]).sum()
input_data_2 = np.array([node_0_value, node_1_value])
output_value = (input_data_2 * weights["output"]).sum()

print(node_0_value, node_1_value)
print(output_value)
```
